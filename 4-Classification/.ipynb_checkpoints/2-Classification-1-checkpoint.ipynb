```python
import pandas as pd

cuisine_df = pd.read_csv("../data/cleaned_cuisine.csv")
cuisine_df.head()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import accuracy_score, precision_score, confusion_matrix, classification_report, precision_recall_curve
from sklearn.svm import SVC
import numpy as np
```

```python
cuisine_label_df = cuisine_df["cuisine"]
cuisine_label_df.head()
```

```python
cuisine_feature_df = cuisine_df.drop(["Unnamed: 0", "cuisine"], axis=1)
cuisine_feature_df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(cuisine_feature_df, cuisine_label_df, test_size=0.3)
```

```python
from sklearn.multiclass import OneVsRestClassifier
model = OneVsRestClassifier(LogisticRegression(solver="liblinear"))
model.fit(X_train, y_train)
# the ravel function flatten data
accuracy = model.score(X_test, y_test)
print(f"accuracy is {accuracy}")
```

```python
print(f"ingredients: {X_test.iloc[50][X_test.iloc[50]!=0].keys()}")
print(f"cuisine: {y_test.iloc[50]}")
```

```python
# fetch a sample then predict it
test = X_test.iloc[50].to_frame().T
proba = model.predict_proba(test)
classes = model.classes_
resultdf = pd.DataFrame(data=proba, columns=classes)

topPrediction = resultdf.T.sort_values(by=[0], ascending = [False])
topPrediction.head()
```

```python
# predict all test set
y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred))
```
